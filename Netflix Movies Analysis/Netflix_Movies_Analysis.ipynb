Netflix Movies Analysis

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("shivamb/netflix-shows")

print("Path to dataset files:", path)
```

```python
# Getting the data
import pandas as pd
data = pd.read_csv('/kaggle/input/netflix-shows/netflix_titles.csv')
```

```python
# Converting the data to a pandas dataframe
df = pd.DataFrame(data)
df.head()
```

```python
# Checking the null values
df.isnull().sum()
```

```python
# Exploring the data
df.info()
```

```python
# Dropping the null values because we have a very large dataset
df.dropna(inplace=True)
```

```python
# Again confirming the dataframe doesn't contains any null values
df.isnull().sum()
```

```python
# Checking the shape of the dataframe
df.shape
```

```python
# Removing leading whitespaces
columns_to_strip = ['type', 'title', 'director', 'cast', 'country', 'rating', 'duration', 'listed_in', 'description']
for column in columns_to_strip:
  df[column] = df[column].str.strip()
```

## Content Composition

Count of Movies vs TV Shows

```python
count_mov_tv = df["type"].value_counts()
count_mov_tv
```

Distribution of titles across release years

```python
dist_title_by_year = df.groupby("release_year")["title"].value_counts()
dist_title_by_year
```

Titles added per year

```python
titles_per_year = df.groupby("release_year")["title"].count().sort_values(ascending=False)
titles_per_year
```

## Geographical Insights

Top 10 countries producing content

```python
top_10_countries = df["country"].str.strip().value_counts().sort_values(ascending=False)
top_10_countries.head(10)
```

Titles by country

```python
df_exploded = df.assign(country=df['country'].str.split(',')).explode('country')
df_exploded['country'] = df_exploded['country'].str.strip()
titles_by_country = df_exploded.groupby("country")["title"].count().sort_values(ascending=False)
titles_by_country
```

Compare diversity of content by region

```python
diversity_by_region = df_exploded.groupby("country")["type"].nunique().sort_values(ascending=False)
diversity_by_region
```

## Genre & Rating Analysis

Most common genres

```python
common_genre = df["listed_in"].str.split(',').explode().str.strip().value_counts().sort_values(ascending=False)
common_genre.head(15)
```

Most common TV ratings

```python
common_ratings = df['rating'].value_counts().sort_values(ascending=False)
common_ratings
```

Titles by rating category

```python
titles_by_rating = df.groupby("rating")["title"].count().sort_values(ascending=False)
titles_by_rating
```

## Release Trends

Titles released over time

```python
titles_over_time = df["release_year"].value_counts().sort_values(ascending=False)
titles_over_time
```

## Director & Cast Insights

Directors with the most titles

```python
dir_with_most_titles = df["director"].str.split(',').explode().str.strip().value_counts().sort_values(ascending=False)
dir_with_most_titles
```

Cast members with most appearances

```python
cast_with_most_app = df['cast'].str.split(',').explode().str.strip().value_counts().sort_values(ascending=False)
cast_with_most_app
```

Most frequent genre combinations

```python
genre_combination = df['listed_in'].str.split(',').explode().str.strip().value_counts().sort_values(ascending=False)
genre_combination.head(15)
```
