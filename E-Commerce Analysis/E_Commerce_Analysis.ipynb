E-Commerce Analysis Using Pandas

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("zusmani/pakistans-largest-ecommerce-dataset")

print("Path to dataset files:", path)
```

```python
# Getting the data
import pandas as pd
data = pd.read_csv('/kaggle/input/pakistans-largest-ecommerce-dataset/Pakistan Largest Ecommerce Dataset.csv')
```

```python
# Converting the data to a pandas dataframe
df = pd.DataFrame(data)
df.head()
```

```python
# Checking the null values
df.isnull().sum()
```

```python
# Exploring the data
df.info()
```

```python
df.describe()
```

```python
# Dropping the unecessary columns
df = df.drop(["sales_commission_code", "FY", "Unnamed: 21", "Unnamed: 22", "Unnamed: 23", "Unnamed: 24", "Unnamed: 25"], axis=1)
df
```

```python
# The dataset contains Item ID, Order Status (Completed, Cancelled, Refund), Date of Order, Stock Keeping Unit, Price, Quantity, Grand Total, Category, Payment Method and Customer ID.
df.head()
```

```python
# Checking the null values now
df.isnull().sum()
```

```python
# Dropping the null values because we have a very large dataset
df.dropna(inplace=True)
```

```python
# Again confirming the dataframe doesn't contains any null values
df.isnull().sum()
```

```python
df.shape
```

```python
df.info()
```

## Sales & Orders Insights

Total Number of Orders

```python
total_orders = len(df)
print("Total Number of Orders:", total_orders)
```

Total Revenue

```python
total_revenue = df["grand_total"].sum()
print("Total Revenue:", total_revenue.round(1))
```

Average Order Value

```python
avg_order_value = total_revenue / total_orders
print("Average Order Value:", avg_order_value.round(1))
```

Orders by Year

```python
yearly_orders = df.groupby(df["Year"]).size().sort_values(ascending=False)
yearly_orders
```

Orders by Months

```python
monthly_orders = df.groupby(df["Month"]).size().sort_values(ascending=False)
monthly_orders
```

Yearly Revenue

```python
yearly_revenue = df.groupby(df["Year"])["grand_total"].sum().sort_values(ascending=False)
yearly_revenue
```

Monthly Revenue

```python
yearly_revenue = df.groupby(df["Month"])["grand_total"].sum().sort_values(ascending=False)
yearly_revenue
```

## Product / Category Analysis

Top Product Categories by Orders

```python
cat_by_orders = df.groupby("category_name_1").size().sort_values(ascending=False)
cat_by_orders
```

Top Product Categories by Revenue

```python
rev_by_cat = df.groupby("category_name_1")["grand_total"].sum().sort_values(ascending=False)
rev_by_cat
```

## Payment Method Analysis

Most popular payment methods

```python
payment_method = df.groupby("payment_method").size().sort_values(ascending=False)
payment_method
```

Revenue by payment method

```python
rev_by_payment_method = df.groupby("payment_method")["grand_total"].sum().sort_values(ascending=False)
rev_by_payment_method.to_frame().style.format('{0:,.0f}')
```

Refund rate by payment method

```python
refunded_order_count = df[df["status"]=='order_refunded'].groupby(df["payment_method"]).size()
refunded_order_rate = (refunded_order_count / refunded_order_count.sum()) * 100
refunded_order_rate.sort_values(ascending=False)
```

Completion rate by payment method

```python
comp_order_count = df[df["status"]=='complete'].groupby(df["payment_method"]).size()
comp_order_rate = (comp_order_count / comp_order_count.sum()) * 100
comp_order_rate.sort_values(ascending=False)
```

## Customer Behavior

Order frequency by customer

```python
order_freq = df.groupby("Customer ID").size().sort_values(ascending=False)
order_freq
```
