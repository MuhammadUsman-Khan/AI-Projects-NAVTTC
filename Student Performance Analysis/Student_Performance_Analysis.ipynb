Now Using all of the Features find the best and the worst student

Best One

```python
index_of_max_score_overall = np.argmax(num_data[:, 4])

max_total_score_overall = num_data[index_of_max_score_overall, 4]

student_id_with_max_score_overall = num_data[index_of_max_score_overall, 0]

best_student_all_features = num_data[index_of_max_score_overall, 1:5]

print(f"The 'best' student (based on highest Total Score) is Student ID: {student_id_with_max_score_overall}")
print(f"Their Total Score is: {max_total_score_overall}")
print(f"Their other performance features (Weekly Self-Study Hours, Attendance Percentage, Class Participation, Total Score) are: {best_student_all_features}")
```

Worst One

```python
index_of_min_score_overall = np.argmin(num_data[:, 4])

min_total_score_overall = num_data[index_of_min_score_overall, 4]

student_id_with_min_score_overall = num_data[index_of_min_score_overall, 0]

worst_student_all_features = num_data[index_of_min_score_overall, 1:5]

print(f"The 'worst' student (based on lowest Total Score) is Student ID: {student_id_with_min_score_overall}")
print(f"Their Total Score is: {min_total_score_overall}")
print(f"Their other performance features (Weekly Self-Study Hours, Attendance Percentage, Class Participation, Total Score) are: {worst_student_all_features}")
```

Correlation between the Columns

```python
numerical_features = num_data[:, 1:5]
correlation_matrix = np.corrcoef(numerical_features, rowvar=False)

print("Correlation Coefficient Matrix for Numerical Features:")
print(correlation_matrix.round(3))

feature_names = ['Weekly Self-Study Hours', 'Attendance Percentage', 'Class Participation', 'Total Score']
print("\nCorresponding Features:")
for i, name in enumerate(feature_names):
    print(f"Column {i}: {name}")
```

This means that:


*   Total Score Increases strongly with the Weekly Self Study Hours and also a bit from the class participation whereas it decreases it a bit from the Attendence Percentage
*   So we can say that the more the weekly self study hours the more score it get
